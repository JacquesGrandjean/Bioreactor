```python
import numpy as np
from scipy.optimize import fsolve

F = 100.0 # l/h
s0 = 10.0 # g/l
Y = 0.5 # g/g
mu_m = 1.0 # h^-1
Ks = 0.75 # g/L

def solve_system(V1, V2):
    D1 = F / V1
    D2 = F / V2
    
    #R1
    if D1 < (mu_m * s0 / (Ks + s0)):
        s1 = (Ks * D1) / (mu_m - D1)
        x1 = Y * (s0 - s1)
    else:
        s1, x1 = s0, 0.0
    
    mu1 = mu_m * s1 / (Ks + s1) if x1 > 0 else 0.0

    #R2
    def equations_R2(vars):
        x2, s2 = vars
        s2_phys = max(s2, 1e-9) 
        mu2 = mu_m * s2_phys / (Ks + s2_phys)
        eq_x = D2 * x1 + (mu2 - D2) * x2
        eq_s = D2 * s1 - (mu2 * x2 / Y) - D2 * s2
        return [eq_x, eq_s]

    # Guess
    x2, s2 = fsolve(equations_R2, [x1 + 4.0, 0.1])
    mu2 = mu_m * max(s2, 0) / (Ks + max(s2, 0))
    
    return (x1, s1, mu1), (x2, s2, mu2)
```

a)

```python
V1a, V2a = 800, 200 #L
r1a, r2a = solve_system(V1a, V2a)
print(f"{'x1':<5} | {'s1':<5} | {'mu1':<5} | {'x2':<5} | {'s2':<5} | {'mu2':<5}")
print(f"{r1a[0]:.3f} | {r1a[1]:.3f} | {r1a[2]:.3f} | {r2a[0]:.3f} | {r2a[1]:.3f} | {r2a[2]:.3f}")
```
